### Exercício 1.2: Filtrando Resultados com `params`

```python
import requests

# URL da API do JSONPlaceholder para posts
url = "https://jsonplaceholder.typicode.com/posts"

# Parâmetros para filtrar os resultados (userId igual a 2)
params = {
    "userId": 2
}

# Realiza a requisição GET com os parâmetros
response = requests.get(url, params=params, timeout=10)

# Exibe o status da requisição e a URL final
print(f"Status Code: {response.status_code}")
print(f"URL Final: {response.url}")

# Exibe o conteúdo da resposta para verificar o filtro
print("Conteúdo (primeiros 500 caracteres) com filtro userId=2:")
print(response.text[:500])
```

## Nível 2: Intermediário (JSON, Erros e Arquivos Binários)

### Exercício 2.1: Consulta de CEPs e Pandas DataFrame

```python
import requests
import pandas as pd

# Lista de CEPs para consultar
ceps = ["01001000", "20040030", "70070100"]

# Lista para armazenar os dados de cada CEP
cep_data = []

# Faz um loop para consultar cada CEP
for cep in ceps:
    url = f"https://viacep.com.br/ws/{cep}/json/"
    print(f"Consultando CEP: {cep}")
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status() # Lança um erro para status HTTP ruins (4xx ou 5xx)
        data = response.json()
        cep_data.append(data)
    except requests.exceptions.RequestException as e:
        print(f"Erro ao consultar o CEP {cep}: {e}")
        cep_data.append({"cep": cep, "erro": str(e)})

# Converte a lista de dicionários em um DataFrame do Pandas
df_ceps = pd.DataFrame(cep_data)

# Exibe o DataFrame
print("\nDataFrame de CEPs:")
print(df_ceps.head())
```

### Exercício 2.2: Função de Download Segura com Tratamento de Erros

```python
import requests

def safe_download(url, timeout=10, headers=None):
    """
    Realiza uma requisição GET de forma segura, tratando erros HTTP.

    Args:
        url (str): A URL para a qual a requisição será feita.
        timeout (int): Tempo limite para a requisição em segundos.
        headers (dict, optional): Dicionário de cabeçalhos HTTP a serem enviados com a requisição.

    Returns:
        requests.Response or None: O objeto de resposta se a requisição for bem-sucedida,
                                   caso contrário, None.
    """
    try:
        response = requests.get(url, timeout=timeout, headers=headers)
        response.raise_for_status()  # Levanta um erro para status de resposta HTTP ruins (4xx ou 5xx)
        print(f"Requisição bem-sucedida para: {url}")
        return response
    except requests.exceptions.HTTPError as e:
        print(f"Erro HTTP ao baixar de {url}: {e}")
        print(f"Status Code: {e.response.status_code}")
        return None
    except requests.exceptions.ConnectionError as e:
        print(f"Erro de conexão ao baixar de {url}: {e}")
        return None
    except requests.exceptions.Timeout as e:
        print(f"Tempo limite excedido ao baixar de {url}: {e}")
        return None
    except requests.exceptions.RequestException as e:
        print(f"Erro inesperado ao baixar de {url}: {e}")
        return None

# Exemplo de uso com uma URL válida
print("\nTestando com URL válida:")
valid_url = "https://jsonplaceholder.typicode.com/posts/1"
response_valid = safe_download(valid_url)
if response_valid:
    print("Conteúdo recebido (primeiros 100 caracteres):", response_valid.text[:100])

# Exemplo de uso com uma URL inválida (que deve gerar um erro 404)
print("\nTestando com URL inválida (espera-se erro 404):")
invalid_url = "https://jsonplaceholder.typicode.com/nonexistent_page"
response_invalid = safe_download(invalid_url)
```

### Exercício 2.3: Download e Salvamento de Imagem Binária

```python
import requests

# A função safe_download já foi definida no Exercício 2.2
# Se estiver executando este notebook do zero, execute o Exercício 2.2 antes.

image_url = "https://picsum.photos/400/400"
file_name = "random_image.jpg"

print(f"\nTentando baixar imagem de: {image_url}")
response_image = safe_download(image_url, timeout=15) # Aumentado timeout para download de arquivos maiores

if response_image:
    try:
        with open(file_name, 'wb') as f:
            f.write(response_image.content)
        print(f"Imagem salva com sucesso como: {file_name}")
    except IOError as e:
        print(f"Erro ao salvar a imagem em arquivo: {e}")
else:
    print("Não foi possível baixar a imagem.")
```

### Exercício 3.2: Webscraping com BeautifulSoup (Títulos e Preços)

```python
import requests
from bs4 import BeautifulSoup

# A função safe_download já foi definida no Exercício 2.2
# Se estiver executando este notebook do zero, execute o Exercício 2.2 antes.

# URL do site para raspagem
scrape_url = "https://books.toscrape.com/"

print(f"\nRealizando requisição para: {scrape_url}")
response_scrape = safe_download(scrape_url)

if response_scrape:
    # Analisa o conteúdo HTML da página
    soup = BeautifulSoup(response_scrape.text, 'html.parser')

    # Encontra todos os artigos de livro (cada um representa um livro)
    books = soup.find_all('article', class_='product_pod')

    print("\nExtraindo Título e Preço dos 5 primeiros livros:")
    # Extrai o título e o preço dos 5 primeiros livros
    for i, book in enumerate(books[:5]): # Pega apenas os 5 primeiros
        title = book.h3.a['title']
        price = book.find('p', class_='price_color').text
        print(f"Livro {i+1}:")
        print(f"  Título: {title}")
        print(f"  Preço: {price}\n")
else:
    print("Não foi possível acessar a página para raspagem.")
```

### Exercício 3.3: Salvando Dados Extraídos em CSV com Pandas

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# A função safe_download já foi definida no Exercício 2.2
# Se estiver executando este notebook do zero, execute o Exercício 2.2 antes.

# URL do site para raspagem
scrape_url = "https://books.toscrape.com/"

print(f"\nRealizando requisição para: {scrape_url}")
response_scrape = safe_download(scrape_url)

if response_scrape:
    # Analisa o conteúdo HTML da página
    soup = BeautifulSoup(response_scrape.text, 'html.parser')

    # Encontra todos os artigos de livro
    books = soup.find_all('article', class_='product_pod')

    # Lista para armazenar os dados dos livros
    book_data = []

    print("\nColetando dados dos livros para o CSV...")
    # Extrai o título e o preço de todos os livros na primeira página
    for book in books:
        title = book.h3.a['title']
        price = book.find('p', class_='price_color').text.replace('Â£', '') # Remove o símbolo da moeda
        book_data.append({"Title": title, "Price (GBP)": float(price)})

    # Cria um DataFrame do Pandas com os dados coletados
    df_books = pd.DataFrame(book_data)

    # Nome do arquivo CSV
    csv_file_name = "books_data.csv"

    # Salva o DataFrame em um arquivo CSV
    try:
        df_books.to_csv(csv_file_name, index=False, encoding='utf-8')
        print(f"Dados dos livros salvos com sucesso em: {csv_file_name}")
        print("\nPrimeiras 5 linhas do DataFrame:")
        print(df_books.head())
    except IOError as e:
        print(f"Erro ao salvar o arquivo CSV: {e}")
else:
    print("Não foi possível acessar a página para coletar os dados dos livros.")
```

### Exercício 3.4: Extração de Tabelas da Wikipedia com `pandas.read_html()`

```python
import pandas as pd
import requests
from io import StringIO

# A função safe_download já foi definida no Exercício 2.2
# Se estiver executando este notebook do zero, execute o Exercício 2.2 antes.

# URL de uma página da Wikipedia com tabelas. Exemplo: Lista de países por população
# Certifique-se de que a URL contenha pelo menos uma tabela HTML.
wikipedia_url = "https://pt.wikipedia.org/wiki/Lista_de_pa%C3%ADses_por_popula%C3%A7%C3%A3o"

# Adicionando um User-Agent para evitar o erro 403 Forbidden da Wikipedia
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
}

print(f"\nTentando baixar o conteúdo de: {wikipedia_url}")
# Passando os cabeçalhos para a função safe_download
response_wiki = safe_download(wikipedia_url, headers=headers)

if response_wiki:
    # Usa io.StringIO para tratar o texto como um arquivo para pandas.read_html
    try:
        # read_html retorna uma lista de DataFrames, um para cada tabela encontrada
        dfs = pd.read_html(StringIO(response_wiki.text))

        print(f"\n{len(dfs)} tabelas encontradas na página. Exibindo a primeira tabela:\n")

        # Exibe a primeira tabela encontrada (geralmente a mais relevante)
        if dfs:
            df_population = dfs[0]
            print(df_population.head())
        else:
            print("Nenhuma tabela encontrada na página.")

    except Exception as e:
        print(f"Erro ao ler tabelas HTML com pandas: {e}")
else:
    print("Não foi possível acessar a página da Wikipedia para extrair tabelas.")
```
